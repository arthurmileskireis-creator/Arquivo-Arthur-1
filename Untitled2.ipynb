```python
from google.colab import userdata
userdata.get('secretName')
# ============================================================
# 🤖 MENTORIA DE CARREIRA COM IA
# Google Colab + OpenRouter + Gradio
# ============================================================

# 1. INSTALAR BIBLIOTECAS
!pip -q install -U openai gradio


# ============================================================
# 2. IMPORTAÇÕES
# ============================================================

import gradio as gr
from openai import OpenAI
from google.colab import userdata


# ============================================================
# 3. CONFIGURAÇÃO DA API
# ============================================================

API_KEY = userdata.get("OPENROUTER_API_KEY")

if not API_KEY:
    raise ValueError(
        """
        ❌ Chave da API não encontrada.

        No Google Colab:
        1. Clique no ícone 🔑 Secrets
        2. Crie OPENROUTER_API_KEY
        3. Cole sua chave do OpenRouter
        4. Execute novamente
        """
    )

client = OpenAI(
    base_url="https://openrouter.ai/api/v1",
    api_key=API_KEY
)

MODEL = "openai/gpt-oss-20b:free"


# ============================================================
# 4. SYSTEM PROMPT
# ============================================================

SYSTEM_PROMPT = """
Você é o MentorIA, um mentor de carreiras baseado em inteligência artificial.

Sua função é ajudar pessoas a explorar possibilidades profissionais,
desenvolver habilidades e criar planos de desenvolvimento.

REGRAS:

- Nunca invente informações sobre o usuário.
- Não prometa emprego, salário ou sucesso.
- Não trate uma carreira como escolha certa ou garantida.
- Apresente alternativas quando houver mais de um caminho possível.
- Explique os motivos das suas sugestões.
- Considere experiência, interesses, habilidades, objetivos e tempo disponível.
- Priorize ações práticas.
- Sugira projetos que possam virar portfólio.
- Não solicite senhas, tokens ou chaves de API.
- Não faça julgamentos discriminatórios.

Quando possível, organize as respostas usando:

🎯 Objetivo
🔎 Perfil
💡 Possibilidades
🧠 Habilidades
📚 Estudos
🛠️ Projetos
📅 Próximos passos
"""


# ============================================================
# 5. SEGURANÇA
# ============================================================

def verificar_entrada(texto):

    termos_sensiveis = [
        "minha senha",
        "senha da minha conta",
        "chave da api",
        "api key",
        "token secreto",
        "senha do banco",
        "cartão de crédito"
    ]

    texto = texto.lower()

    for termo in termos_sensiveis:
        if termo in texto:
            return False

    return True


# ============================================================
# 6. FUNÇÃO PRINCIPAL DA IA
# ============================================================

def chamar_ia(mensagem, contexto=""):

    if not verificar_entrada(mensagem):
        return (
            "⚠️ Por segurança, não compartilhe senhas, "
            "tokens, chaves de API ou dados financeiros."
        )

    prompt = f"""
CONTEXTO DO USUÁRIO:

{contexto}

MENSAGEM:

{mensagem}
"""

    try:

        response = client.responses.create(
            model=MODEL,
            instructions=SYSTEM_PROMPT,
            input=prompt
        )

        return response.output_text

    except Exception as erro:

        return f"""
❌ Não foi possível consultar a IA.

Verifique:
- sua chave do OpenRouter;
- sua conexão;
- o modelo configurado.

Detalhes técnicos:
{erro}
"""


# ============================================================
# 7. CRIAR PERFIL
# ============================================================

def criar_contexto(nome, experiencia, interesses, habilidades, objetivo, tempo):

    return f"""
Nome: {nome}

Experiência profissional:
{experiencia}

Interesses:
{interesses}

Habilidades atuais:
{habilidades}

Objetivo profissional:
{objetivo}

Tempo disponível para estudar:
{tempo}
"""


# ============================================================
# 8. ANÁLISE DE CARREIRA
# ============================================================

def analisar_carreira(nome, experiencia, interesses, habilidades, objetivo, tempo):

    contexto = criar_contexto(
        nome,
        experiencia,
        interesses,
        habilidades,
        objetivo,
        tempo
    )

    pergunta = """
Analise meu perfil profissional.

Apresente:

1. 🎯 Meu objetivo
2. 🔎 Um resumo do meu perfil
3. 💡 Possíveis caminhos profissionais
4. 🧠 Habilidades que já parecem relevantes
5. 📚 Habilidades que preciso desenvolver
6. 🛠️ Projetos que poderiam fortalecer meu portfólio
7. 📅 Próximos passos

Não escolha uma única carreira como se fosse uma certeza.
Apresente alternativas e explique os critérios utilizados.
"""

    return chamar_ia(pergunta, contexto)


# ============================================================
# 9. PLANO DE ESTUDOS
# ============================================================

def criar_plano(nome, experiencia, interesses, habilidades, objetivo, tempo):

    contexto = criar_contexto(
        nome,
        experiencia,
        interesses,
        habilidades,
        objetivo,
        tempo
    )

    pergunta = """
Crie um plano de desenvolvimento profissional personalizado.

Divida em:

FASE 1 — Fundamentos
FASE 2 — Prática
FASE 3 — Projetos
FASE 4 — Portfólio
FASE 5 — Preparação profissional

Para cada fase informe:

- O que estudar
- O que praticar
- Um projeto sugerido
- Resultado esperado

Considere o tempo disponível informado no perfil.

Se possível, organize também uma sugestão de cronograma de 30, 60
e 90 dias.
"""

    return chamar_ia(pergunta, contexto)


# ============================================================
# 10. PROJETOS PARA PORTFÓLIO
# ============================================================

def gerar_projetos(nome, experiencia, interesses, habilidades, objetivo, tempo):

    contexto = criar_contexto(
        nome,
        experiencia,
        interesses,
        habilidades,
        objetivo,
        tempo
    )

    pergunta = """
Sugira 5 projetos para meu portfólio profissional.

Para cada projeto informe:

🛠️ Nome do projeto
🎯 Objetivo
🧠 Habilidades desenvolvidas
💻 Tecnologias possíveis
📈 Nível de dificuldade
📋 O que eu deveria entregar
💼 Como esse projeto poderia demonstrar minhas habilidades

Comece com projetos compatíveis com meu nível atual e aumente
gradualmente a dificuldade.
"""

    return chamar_ia(pergunta, contexto)


# ============================================================
# 11. CHAT
# ============================================================

def responder_chat(
    mensagem,
    historico,
    nome,
    experiencia,
    interesses,
    habilidades,
    objetivo,
    tempo
):

    contexto = criar_contexto(
        nome,
        experiencia,
        interesses,
        habilidades,
        objetivo,
        tempo
    )

    resposta = chamar_ia(mensagem, contexto)

    return resposta


# ============================================================
# 12. INTERFACE
# ============================================================

CSS = """
.gradio-container {
    max-width: 1100px !important;
    margin: auto;
}

.titulo {
    text-align: center;
}

.subtitulo {
    text-align: center;
    color: #666;
}
"""


with gr.Blocks(
    theme=gr.themes.Soft(
        primary_hue="blue",
        secondary_hue="indigo"
    ),
    css=CSS,
    title="MentorIA"
) as app:

    gr.Markdown(
        """
        # 🤖 MentorIA
        ### Seu mentor de desenvolvimento profissional com Inteligência Artificial

        Explore possibilidades de carreira, descubra habilidades,
        crie projetos e monte seu plano de desenvolvimento.
        """,
        elem_classes="titulo"
    )

    # --------------------------------------------------------
    # PERFIL
    # --------------------------------------------------------

    with gr.Tab("👤 Meu Perfil"):

        gr.Markdown(
            """
            ## Conte um pouco sobre você

            Quanto mais informações você fornecer, mais personalizada
            poderá ser a análise.
            """
        )

        nome = gr.Textbox(
            label="Seu nome",
            placeholder="Ex.: João"
        )

        experiencia = gr.Textbox(
            label="Experiência profissional",
            placeholder="Conte brevemente onde já trabalhou ou estudou.",
            lines=3
        )

        interesses = gr.Textbox(
            label="Seus interesses",
            placeholder="Ex.: tecnologia, negócios, programação, design...",
            lines=3
        )

        habilidades = gr.Textbox(
            label="Habilidades atuais",
            placeholder="Ex.: Excel, Python básico, comunicação...",
            lines=3
        )

        objetivo = gr.Textbox(
            label="Seu objetivo profissional",
            placeholder="Ex.: conseguir meu primeiro emprego em tecnologia.",
            lines=3
        )

        tempo = gr.Textbox(
            label="Tempo disponível para estudar",
            placeholder="Ex.: 1 hora por dia"
        )


    # --------------------------------------------------------
    # ANÁLISE
    # --------------------------------------------------------

    with gr.Tab("🎯 Análise de Carreira"):

        botao_analise = gr.Button(
            "🔎 Analisar meu perfil",
            variant="primary"
        )

        resultado_analise = gr.Markdown(
            "Sua análise aparecerá aqui."
        )

        botao_analise.click(
            fn=analisar_carreira,
            inputs=[
                nome,
                experiencia,
                interesses,
                habilidades,
                objetivo,
                tempo
            ],
            outputs=resultado_analise
        )


    # --------------------------------------------------------
    # PLANO
    # --------------------------------------------------------

    with gr.Tab("📚 Plano de Estudos"):

        gr.Markdown(
            """
            ## 📅 Seu plano personalizado

            A IA criará um roteiro progressivo baseado no seu perfil.
            """
        )

        botao_plano = gr.Button(
            "🚀 Criar meu plano",
            variant="primary"
        )

        resultado_plano = gr.Markdown(
            "Seu plano aparecerá aqui."
        )

        botao_plano.click(
            fn=criar_plano,
            inputs=[
                nome,
                experiencia,
                interesses,
                habilidades,
                objetivo,
                tempo
            ],
            outputs=resultado_plano
        )


    # --------------------------------------------------------
    # PROJETOS
    # --------------------------------------------------------

    with gr.Tab("🛠️ Projetos"):

        gr.Markdown(
            """
            ## 💼 Projetos para seu portfólio

            Transforme o aprendizado em projetos práticos.
            """
        )

        botao_projetos = gr.Button(
            "💡 Gerar projetos",
            variant="primary"
        )

        resultado_projetos = gr.Markdown(
            "As sugestões aparecerão aqui."
        )

        botao_projetos.click(
            fn=gerar_projetos,
            inputs=[
                nome,
                experiencia,
                interesses,
                habilidades,
                objetivo,
                tempo
            ],
            outputs=resultado_projetos
        )


    # --------------------------------------------------------
    # CHAT
    # --------------------------------------------------------

    with gr.Tab("💬 Conversar com o Mentor"):

        gr.Markdown(
            """
            ## 🤖 Converse com seu MentorIA

            Pergunte sobre estudos, programação, carreira,
            projetos, portfólio e desenvolvimento profissional.
            """
        )

        chatbot = gr.Chatbot(
            label="MentorIA",
            height=450
        )

        mensagem = gr.Textbox(
            label="Sua mensagem",
            placeholder="Digite sua pergunta...",
            lines=3
        )

        enviar = gr.Button(
            "Enviar 🚀",
            variant="primary"
        )

        def chat(
            mensagem,
            historico,
            nome,
            experiencia,
            interesses,
            habilidades,
            objetivo,
            tempo
        ):

            if not mensagem.strip():
                return historico, ""

            contexto = criar_contexto(
                nome,
                experiencia,
                interesses,
                habilidades,
                objetivo,
                tempo
            )

            resposta = chamar_ia(
                mensagem,
                contexto
            )

            historico = historico or []

            historico.append({
                "role": "user",
                "content": mensagem
            })

            historico.append({
                "role": "assistant",
                "content": resposta
            })

            return historico, ""

        enviar.click(
            fn=chat,
            inputs=[
                mensagem,
                chatbot,
                nome,
                experiencia,
                interesses,
                habilidades,
                objetivo,
                tempo
            ],
            outputs=[
                chatbot,
                mensagem
            ]
        )

        mensagem.submit(
            fn=chat,
            inputs=[
                mensagem,
                chatbot,
                nome,
                experiencia,
                interesses,
                habilidades,
                objetivo,
                tempo
            ],
            outputs=[
                chatbot,
                mensagem
            ]
        )


    # --------------------------------------------------------
    # RODAPÉ
    # --------------------------------------------------------

    gr.Markdown(
        """
        ---

        ### 🚀 MentorIA

        Projeto educacional de programação com Inteligência Artificial.

        **Tecnologias:** Python · Google Colab · Gradio · OpenRouter · IA

        ⚠️ As sugestões do MentorIA são informativas e não garantem
        resultados profissionais.
        """
    )


# ============================================================
# 13. EXECUTAR A APLICAÇÃO
# ============================================================

app.launch(
    share=True,
    debug=True
)
```
