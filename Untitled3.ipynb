```python
%%html

<!DOCTYPE html>
<html lang="pt-BR">
<head>
    <meta charset="UTF-8">
    <meta name="viewport" content="width=device-width, initial-scale=1.0">

    <title>Aventura na Floresta</title>

    <style>
        /* =========================
           ESTILO GERAL
        ========================= */

        * {
            box-sizing: border-box;
            margin: 0;
            padding: 0;
        }

        body {
            font-family: Arial, Helvetica, sans-serif;
            background: #071b12;
            color: white;
            min-height: 100vh;
            display: flex;
            justify-content: center;
            align-items: center;
            padding: 20px;
        }

        /* =========================
           CONTAINER DA AVENTURA
        ========================= */

        .aventura {
            width: 100%;
            max-width: 800px;
            background: #123524;
            border-radius: 20px;
            overflow: hidden;
            box-shadow: 0 10px 30px rgba(0, 0, 0, 0.5);
        }

        /* =========================
           CABEÇALHO
        ========================= */

        .cabecalho {
            background: #1e6f46;
            padding: 25px;
            text-align: center;
        }

        .cabecalho h1 {
            font-size: 32px;
            margin-bottom: 8px;
        }

        .cabecalho p {
            color: #d8f3dc;
        }

        /* =========================
           IMAGEM
        ========================= */

        .imagem-aventura {
            width: 100%;
            height: 300px;
            object-fit: cover;
            display: block;
        }

        /* =========================
           HISTÓRIA
        ========================= */

        .conteudo {
            padding: 30px;
        }

        #texto-aventura {
            font-size: 20px;
            line-height: 1.6;
            margin-bottom: 25px;
            text-align: center;
        }

        /* =========================
           BOTÕES
        ========================= */

        .opcoes {
            display: flex;
            flex-direction: column;
            gap: 15px;
        }

        .botao {
            border: none;
            padding: 16px;
            border-radius: 10px;
            background: #52b788;
            color: white;
            font-size: 17px;
            font-weight: bold;
            cursor: pointer;
            transition: 0.3s;
        }

        .botao:hover {
            background: #74c69d;
            transform: translateY(-2px);
        }

        .botao:active {
            transform: scale(0.98);
        }

        /* =========================
           STATUS
        ========================= */

        #status {
            margin-top: 25px;
            padding: 15px;
            border-radius: 10px;
            background: #081c15;
            text-align: center;
            color: #b7e4c7;
        }

        /* =========================
           RODAPÉ
        ========================= */

        .rodape {
            text-align: center;
            padding: 20px;
            background: #0b2418;
            color: #95d5b2;
            font-size: 14px;
        }

        /* =========================
           RESPONSIVIDADE
        ========================= */

        @media (max-width: 600px) {

            .cabecalho h1 {
                font-size: 25px;
            }

            .imagem-aventura {
                height: 220px;
            }

            .conteudo {
                padding: 20px;
            }

            #texto-aventura {
                font-size: 17px;
            }
        }
    </style>
</head>

<body>

    <main class="aventura">

        <!-- CABEÇALHO -->
        <header class="cabecalho">
            <h1>🌲 A Floresta Misteriosa</h1>
            <p>Uma aventura interativa</p>
        </header>

        <!-- IMAGEM -->
        <img
            id="imagem-aventura"
            class="imagem-aventura"
            src="https://images.unsplash.com/photo-1448375240586-882707db888b?auto=format&fit=crop&w=1200&q=80"
            alt="Floresta misteriosa"
        >

        <!-- CONTEÚDO -->
        <section class="conteudo">

            <p id="texto-aventura">
                Você está caminhando por uma floresta misteriosa.
                O sol está desaparecendo e você encontra dois caminhos.
                Qual deles você irá escolher?
            </p>

            <!--
                Os atributos data-* guardam informações
                que serão utilizadas pelo JavaScript.
            -->
            <div class="opcoes" id="opcoes">

                <button
                    class="botao"
                    data-escolha="esquerda">
                    🌲 Entrar na floresta
                </button>

                <button
                    class="botao"
                    data-escolha="direita">
                    🏰 Seguir até o castelo
                </button>

            </div>

            <div id="status">
                ❤️ Vida: 3 | ⭐ Pontos: 0
            </div>

        </section>

        <footer class="rodape">
            Aventura criada com HTML, CSS e JavaScript
        </footer>

    </main>

    <script>

        /* =========================================
           VARIÁVEIS DO JOGO
        ========================================= */

        let vida = 3;
        let pontos = 0;

        /*
         * Pegamos elementos HTML através do DOM.
         */

        const texto = document.getElementById("texto-aventura");
        const imagem = document.getElementById("imagem-aventura");
        const opcoes = document.getElementById("opcoes");
        const status = document.getElementById("status");


        /* =========================================
           FUNÇÃO PARA ATUALIZAR O STATUS
        ========================================= */

        function atualizarStatus() {

            status.textContent =
                `❤️ Vida: ${vida} | ⭐ Pontos: ${pontos}`;

        }


        /* =========================================
           FUNÇÃO PARA CRIAR BOTÕES
        ========================================= */

        function criarBotao(textoBotao, escolha) {

            const botao = document.createElement("button");

            botao.classList.add("botao");

            botao.textContent = textoBotao;

            /*
             * Criamos um atributo data-escolha.
             */

            botao.dataset.escolha = escolha;

            /*
             * addEventListener detecta quando
             * o usuário clica no botão.
             */

            botao.addEventListener("click", function() {

                escolher(escolha);

            });

            opcoes.appendChild(botao);
        }


        /* =========================================
           FUNÇÃO PRINCIPAL DA AVENTURA
        ========================================= */

        function escolher(escolha) {

            opcoes.innerHTML = "";

            /* CAMINHO DA FLORESTA */

            if (escolha === "esquerda") {

                pontos += 10;

                texto.textContent =
                    "Você entrou na floresta e encontrou uma pequena cabana abandonada. Dentro dela existe uma chave dourada.";

                imagem.src =
                    "https://images.unsplash.com/photo-1500534623283-312aade485b7?auto=format&fit=crop&w=1200&q=80";

                criarBotao(
                    "🔑 Pegar a chave",
                    "chave"
                );

                criarBotao(
                    "🚶 Continuar pela floresta",
                    "floresta"
                );

            }

            /* CAMINHO DO CASTELO */

            else if (escolha === "direita") {

                pontos += 5;

                texto.textContent =
                    "Você chegou diante de um enorme castelo. O portão está fechado e existe um caminho escuro ao lado.";

                imagem.src =
                    "https://images.unsplash.com/photo-1518709268805-4e9042af9f23?auto=format&fit=crop&w=1200&q=80";

                criarBotao(
                    "🏰 Tentar abrir o portão",
                    "portao"
                );

                criarBotao(
                    "🌑 Entrar no caminho escuro",
                    "escuro"
                );

            }

            /* PEGAR A CHAVE */

            else if (escolha === "chave") {

                pontos += 20;

                texto.textContent =
                    "Você pegou a chave dourada! Ela parece ser muito antiga. Talvez ela possa abrir alguma porta misteriosa.";

                imagem.src =
                    "https://images.unsplash.com/photo-1550745165-9bc0b252726f?auto=format&fit=crop&w=1200&q=80";

                criarBotao(
                    "🏰 Ir para o castelo",
                    "castelo"
                );

                criarBotao(
                    "🌲 Continuar explorando",
                    "floresta"
                );

            }

            /* CONTINUAR NA FLORESTA */

            else if (escolha === "floresta") {

                vida--;

                texto.textContent =
                    "Você caminhou por muito tempo e acabou encontrando uma criatura misteriosa.";

                imagem.src =
                    "https://images.unsplash.com/photo-1473448912268-2022ce9509d8?auto=format&fit=crop&w=1200&q=80";

                criarBotao(
                    "⚔️ Enfrentar a criatura",
                    "monstro"
                );

                criarBotao(
                    "🏃 Fugir",
                    "fugir"
                );

            }

            /* PORTÃO */

            else if (escolha === "portao") {

                vida--;

                texto.textContent =
                    "Você tentou abrir o portão, mas ele estava trancado. Um barulho assustador veio de dentro do castelo.";

                criarBotao(
                    "🔙 Voltar",
                    "inicio"
                );

                criarBotao(
                    "👻 Entrar pela janela",
                    "janela"
                );

            }

            /* CAMINHO ESCURO */

            else if (escolha === "escuro") {

                vida--;

                texto.textContent =
                    "O caminho estava cheio de armadilhas. Você conseguiu escapar, mas perdeu energia.";

                criarBotao(
                    "🔙 Voltar para o castelo",
                    "direita"
                );

                criarBotao(
                    "🌲 Continuar",
                    "floresta"
                );

            }

            /* CASTELO */

            else if (escolha === "castelo") {

                pontos += 30;

                texto.textContent =
                    "A chave abriu uma pequena porta escondida no castelo! Você encontrou um enorme tesouro. 🎉";

                imagem.src =
                    "https://images.unsplash.com/photo-1579621970563-ebec7560ff3e?auto=format&fit=crop&w=1200&q=80";

                finalizar("🏆 VOCÊ ENCONTROU O TESOURO!");

            }

            /* MONSTRO */

            else if (escolha === "monstro") {

                pontos += 50;

                texto.textContent =
                    "Você enfrentou a criatura e descobriu que ela era apenas o guardião de um tesouro escondido!";

                finalizar("🏆 VOCÊ VENCEU A AVENTURA!");

            }

            /* FUGIR */

            else if (escolha === "fugir") {

                texto.textContent =
                    "Você conseguiu fugir e encontrou o caminho de volta para casa.";

                finalizar("🏠 VOCÊ SOBREVIVEU!");

            }

            /* JANELA */

            else if (escolha === "janela") {

                pontos += 15;

                texto.textContent =
                    "Você entrou pela janela e encontrou uma sala secreta cheia de mapas antigos.";

                criarBotao(
                    "🗺️ Explorar a sala",
                    "sala"
                );

            }

            /* SALA */

            else if (escolha === "sala") {

                pontos += 25;

                texto.textContent =
                    "Os mapas revelaram a localização de um tesouro escondido! Você completou sua missão.";

                finalizar("🏆 MISSÃO CONCLUÍDA!");

            }

            /* VOLTAR AO INÍCIO */

            else if (escolha === "inicio") {

                reiniciar();

            }

            atualizarStatus();

            /*
             * Se a vida chegar a zero,
             * o jogador perde.
             */

            if (vida <= 0) {

                texto.textContent =
                    "Você ficou sem energia para continuar sua aventura.";

                finalizar("💀 FIM DE JOGO");

            }

        }


        /* =========================================
           FINAL DA AVENTURA
        ========================================= */

        function finalizar(mensagem) {

            opcoes.innerHTML = "";

            texto.textContent +=
                `\n\n${mensagem} Pontuação final: ${pontos}`;

            criarBotao(
                "🔄 Jogar novamente",
                "inicio"
            );

            atualizarStatus();

        }


        /* =========================================
           REINICIAR JOGO
        ========================================= */

        function reiniciar() {

            vida = 3;
            pontos = 0;

            texto.textContent =
                "Você está caminhando por uma floresta misteriosa. O sol está desaparecendo e você encontra dois caminhos. Qual deles você irá escolher?";

            imagem.src =
                "https://images.unsplash.com/photo-1448375240586-882707db888b?auto=format&fit=crop&w=1200&q=80";

            opcoes.innerHTML = "";

            criarBotao(
                "🌲 Entrar na floresta",
                "esquerda"
            );

            criarBotao(
                "🏰 Seguir até o castelo",
                "direita"
            );

            atualizarStatus();

        }


        /* =========================================
           FOREACH
        ========================================= */

        /*
         * Selecionamos todos os botões existentes.
         * forEach passa por cada botão e adiciona
         * um evento de clique.
         */

        const botoes = document.querySelectorAll(".botao");

        botoes.forEach(function(botao) {

            botao.addEventListener("click", function() {

                const escolha =
                    botao.dataset.escolha;

                escolher(escolha);

            });

        });

    </script>

</body>
</html>

```
