```python
# ============================================================
# ESTATÍSTICA - QUALIDADE DE VIDA DA TURMA 2C
# 22 ALUNOS
# ============================================================

# 1. IMPORTANDO AS BIBLIOTECAS
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# ============================================================
# 2. DADOS DA TURMA
# ============================================================

# 12 alunos = qualidade alta
# 8 alunos  = qualidade média
# 2 alunos  = qualidade baixa

categorias = ["Alta", "Média", "Baixa"]
quantidades = [12, 8, 2]

total_alunos = sum(quantidades)

print("======================================")
print("QUALIDADE DE VIDA - TURMA 2C")
print("======================================")
print("Total de alunos:", total_alunos)
print("Qualidade alta:", quantidades[0])
print("Qualidade média:", quantidades[1])
print("Qualidade baixa:", quantidades[2])

# ============================================================
# 3. CALCULANDO AS PORCENTAGENS
# ============================================================

porcentagens = [
    quantidade / total_alunos * 100
    for quantidade in quantidades
]

print("\nPorcentagens:")
print(f"Alta: {porcentagens[0]:.1f}%")
print(f"Média: {porcentagens[1]:.1f}%")
print(f"Baixa: {porcentagens[2]:.1f}%")

# ============================================================
# 4. PRIMEIRO GRÁFICO - GRÁFICO DE BARRAS
# ============================================================

plt.figure(figsize=(8, 5))

cores = ["green", "orange", "red"]

plt.bar(categorias, quantidades, color=cores, edgecolor="black")

plt.title("Qualidade de Vida da Turma 2C")
plt.xlabel("Nível de qualidade de vida")
plt.ylabel("Número de alunos")

# Mostrar os números em cima das barras
for i, valor in enumerate(quantidades):
    plt.text(i, valor + 0.2, str(valor),
             ha="center", fontsize=12, fontweight="bold")

plt.ylim(0, 14)
plt.grid(axis="y", alpha=0.3)

plt.show()

# ============================================================
# 5. GRÁFICO DE PIZZA
# ============================================================

plt.figure(figsize=(7, 7))

plt.pie(
    quantidades,
    labels=categorias,
    colors=cores,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Distribuição da Qualidade de Vida - Turma 2C")

plt.show()

# ============================================================
# 6. TRANSFORMANDO AS CATEGORIAS EM NÚMEROS
# ============================================================

# Para fazer média, mediana e desvio padrão,
# precisamos transformar as categorias em valores numéricos.

# Baixa = 1
# Média = 2
# Alta = 3

dados = (
    [3] * 12 +   # 12 alunos com qualidade alta
    [2] * 8 +    # 8 alunos com qualidade média
    [1] * 2      # 2 alunos com qualidade baixa
)

print("\nDados numéricos:")
print(dados)

# ============================================================
# 7. MÉDIA
# ============================================================

media = np.mean(dados)

print("\n======================================")
print("ESTATÍSTICA DESCRITIVA")
print("======================================")

print(f"Média: {media:.2f}")

# ============================================================
# 8. MEDIANA
# ============================================================

mediana = np.median(dados)

print(f"Mediana: {mediana:.2f}")

# ============================================================
# 9. DESVIO PADRÃO
# ============================================================

desvio_padrao = np.std(dados)

print(f"Desvio padrão: {desvio_padrao:.2f}")

# ============================================================
# 10. MODA
# ============================================================

valores, frequencias = np.unique(dados, return_counts=True)

indice_moda = np.argmax(frequencias)

moda = valores[indice_moda]

print(f"Moda: {moda}")

# ============================================================
# 11. INTERPRETANDO OS RESULTADOS
# ============================================================

print("\n======================================")
print("INTERPRETAÇÃO")
print("======================================")

print(f"""
A turma 2C possui {total_alunos} alunos.

- {quantidades[0]} alunos ({porcentagens[0]:.1f}%) possuem
  qualidade de vida alta.

- {quantidades[1]} alunos ({porcentagens[1]:.1f}%) possuem
  qualidade de vida média.

- {quantidades[2]} alunos ({porcentagens[2]:.1f}%) possuem
  qualidade de vida baixa.

A média numérica foi {media:.2f}.
A mediana foi {mediana:.2f}.
O desvio padrão foi {desvio_padrao:.2f}.

A categoria mais frequente é a qualidade de vida alta.
""")

# ============================================================
# 12. HISTOGRAMA
# ============================================================

plt.figure(figsize=(8, 5))

plt.hist(
    dados,
    bins=[0.5, 1.5, 2.5, 3.5],
    color="skyblue",
    edgecolor="black"
)

plt.xticks(
    [1, 2, 3],
    ["Baixa", "Média", "Alta"]
)

plt.xlabel("Qualidade de vida")
plt.ylabel("Número de alunos")
plt.title("Histograma da Qualidade de Vida - Turma 2C")

plt.grid(axis="y", alpha=0.3)

plt.show()

# ============================================================
# 13. SIMULAÇÃO DE UMA DISTRIBUIÇÃO GAUSSIANA
# ============================================================

np.random.seed(42)

# Vamos criar 10.000 valores simulados
# usando a média e o desvio padrão encontrados na turma.

dados_simulados = np.random.normal(
    loc=media,
    scale=desvio_padrao,
    size=10000
)

# ============================================================
# 14. HISTOGRAMA + CURVA GAUSSIANA
# ============================================================

plt.figure(figsize=(10, 6))

# Histograma dos dados simulados
plt.hist(
    dados_simulados,
    bins=50,
    density=True,
    alpha=0.6,
    color="lightblue",
    edgecolor="black",
    label="Dados simulados"
)

# Criando os pontos da curva
x = np.linspace(
    media - 4 * desvio_padrao,
    media + 4 * desvio_padrao,
    500
)

# Curva normal
y = norm.pdf(
    x,
    media,
    desvio_padrao
)

plt.plot(
    x,
    y,
    color="red",
    linewidth=3,
    label="Curva Gaussiana"
)

plt.axvline(
    media,
    color="black",
    linestyle="--",
    linewidth=2,
    label=f"Média = {media:.2f}"
)

plt.xlabel("Qualidade de vida")
plt.ylabel("Densidade")
plt.title("Simulação de uma Distribuição Gaussiana")
plt.legend()
plt.grid(alpha=0.3)

plt.show()

# ============================================================
# 15. COMPARAÇÃO ENTRE A TURMA E A SIMULAÇÃO
# ============================================================

media_simulada = np.mean(dados_simulados)
desvio_simulado = np.std(dados_simulados)

print("\n======================================")
print("COMPARAÇÃO")
print("======================================")

print(f"Média da turma:       {media:.2f}")
print(f"Média simulada:       {media_simulada:.2f}")

print(f"\nDesvio da turma:      {desvio_padrao:.2f}")
print(f"Desvio simulado:      {desvio_simulado:.2f}")

# ============================================================
# 16. CONCLUSÃO AUTOMÁTICA
# ============================================================

print("\n======================================")
print("CONCLUSÃO")
print("======================================")

print("""
A análise mostra que a maioria dos alunos da turma 2C
possui qualidade de vida alta.

Dos 22 alunos analisados, 12 estão na categoria alta,
8 estão na categoria média e 2 estão na categoria baixa.

A média, a mediana e o desvio padrão ajudam a resumir
os dados e observar a distribuição das respostas.

A simulação Gaussiana permite visualizar como os dados
poderiam se comportar em uma população maior.

Entretanto, a distribuição Gaussiana é apenas um modelo
matemático. Ela não representa perfeitamente a realidade,
pois a qualidade de vida depende de vários fatores e a
turma possui apenas 22 alunos.

Portanto, os resultados descrevem especificamente a turma
2C e não devem ser generalizados para toda a população.
""")

```
