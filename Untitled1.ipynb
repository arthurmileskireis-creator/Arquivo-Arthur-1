```python
import hashlib
import itertools
import string


# ------------------------------------------------------------
# 2. FUNÇÃO HASH
# ------------------------------------------------------------

def gerar_hash(senha):
    """
    Recebe uma senha e retorna o seu hash SHA-256.
    """

    # Transformamos a senha em bytes
    senha_bytes = senha.encode("utf-8")

    # Geramos o hash usando SHA-256
    hash_senha = hashlib.sha256(senha_bytes).hexdigest()

    return hash_senha


# ------------------------------------------------------------
# 3. CADASTRO DE USUÁRIO
# ------------------------------------------------------------

usuario = "aluno"

# Senha utilizada apenas para a demonstração
senha_correta = "abc"

# Em vez de guardar a senha diretamente,
# guardamos somente o seu hash.
hash_armazenado = gerar_hash(senha_correta)

print("========================================")
print("      CADASTRO DO USUÁRIO")
print("========================================")
print("Usuário:", usuario)
print("Hash armazenado:", hash_armazenado)


# ------------------------------------------------------------
# 4. AUTENTICAÇÃO SEGURA
# ------------------------------------------------------------

def autenticar(usuario_digitado, senha_digitada):
    """
    Verifica se o usuário e a senha estão corretos.
    """

    # Verifica o nome do usuário
    if usuario_digitado != usuario:
        return False

    # Gera o hash da senha que o usuário digitou
    hash_digitado = gerar_hash(senha_digitada)

    # Compara o hash digitado com o hash armazenado
    if hash_digitado == hash_armazenado:
        return True
    else:
        return False


print("\n========================================")
print("          AUTENTICAÇÃO")
print("========================================")

usuario_teste = input("Digite o usuário: ")
senha_teste = input("Digite a senha: ")

if autenticar(usuario_teste, senha_teste):
    print("Autenticação realizada com sucesso!")
else:
    print("Usuário ou senha incorretos.")


# ------------------------------------------------------------
# 5. CÁLCULO DE COMBINAÇÕES
# ------------------------------------------------------------

print("\n========================================")
print("          COMBINAÇÕES")
print("========================================")

# Exemplo:
# Temos 3 caracteres disponíveis: a, b e c
# Queremos formar senhas com 3 caracteres.

caracteres = "abc"
tamanho = 3

# Calculamos todas as combinações possíveis
quantidade_combinacoes = len(caracteres) ** tamanho

print("Caracteres disponíveis:", caracteres)
print("Tamanho da senha:", tamanho)
print("Quantidade de combinações possíveis:",
      quantidade_combinacoes)


# ------------------------------------------------------------
# 6. GERANDO AS COMBINAÇÕES
# ------------------------------------------------------------

print("\nCombinações possíveis:")

contador = 0

for combinacao in itertools.product(caracteres, repeat=tamanho):

    # Transformamos a combinação em uma string
    tentativa = "".join(combinacao)

    print(tentativa)

    contador += 1

print("Total gerado:", contador)


# ------------------------------------------------------------
# 7. ATAQUE DE FORÇA BRUTA
# ------------------------------------------------------------

print("\n========================================")
print("        ATAQUE DE FORÇA BRUTA")
print("========================================")

# Para a demonstração, vamos tentar descobrir a senha
# comparando o hash de cada tentativa com o hash armazenado.

caracteres = "abc"
tamanho_maximo = 3

tentativas = 0
senha_encontrada = None


# O primeiro loop define o tamanho da tentativa.
for tamanho in range(1, tamanho_maximo + 1):

    # O segundo loop gera todas as possibilidades
    # para aquele tamanho.
    for combinacao in itertools.product(caracteres,
                                        repeat=tamanho):

        tentativa = "".join(combinacao)

        tentativas += 1

        # Calculamos o hash da tentativa
        hash_tentativa = gerar_hash(tentativa)

        # Comparamos com o hash verdadeiro
        if hash_tentativa == hash_armazenado:

            senha_encontrada = tentativa

            print("Senha encontrada!")
            print("Senha:", senha_encontrada)
            print("Número de tentativas:", tentativas)

            # Interrompe o loop interno
            break

    # Se encontramos a senha, também encerramos o loop externo.
    if senha_encontrada is not None:
        break


# ------------------------------------------------------------
# 8. RESULTADO DO ATAQUE
# ------------------------------------------------------------

if senha_encontrada is None:

    print("A senha não foi encontrada.")

else:

    print("\n========================================")
    print("          RESULTADO FINAL")
    print("========================================")
    print("Senha descoberta:", senha_encontrada)
    print("Tentativas realizadas:", tentativas)


# ------------------------------------------------------------
# 9. RESUMO DOS CONCEITOS
# ------------------------------------------------------------

print("\n========================================")
print("             RESUMO")
print("========================================")

print("""
1. CRIPTOGRAFIA:
   É utilizada para proteger informações.

2. HASH:
   A senha é transformada em uma sequência de caracteres
   por meio de uma função hash.

3. AUTENTICAÇÃO:
   O programa compara o hash da senha digitada com o
   hash armazenado.

4. COMBINAÇÕES:
   O programa gera diferentes possibilidades de caracteres.

5. FORÇA BRUTA:
   O programa testa várias possibilidades até encontrar
   uma que produza o mesmo hash da senha.

6. IF/ELSE:
   São utilizados para tomar decisões no programa.

7. FOR:
   É utilizado para repetir ações e percorrer as
   possibilidades de senha.

8. FUNÇÕES:
   Foram criadas funções para organizar o código,
   como gerar_hash() e autenticar().
""")
```
